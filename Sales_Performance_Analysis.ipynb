```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/superstore_sales.csv", encoding="latin1")
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'])
```

```python
total_sales = df['Sales'].sum()
total_sales
```

```python
total_orders = df['Order_ID'].nunique()
total_orders
```

```python
total_customers = df['Customer_ID'].nunique()
total_customers
```

```python
total_products = df['Product_ID'].nunique()
total_products
```

```python
average_order_value = total_sales / total_orders
average_order_value
```

```python
kpi_summary = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Orders',
        'Total Customers',
        'Total Products',
        'Average Order Value'
    ],
    'Value': [
        total_sales,
        total_orders,
        total_customers,
        total_products,
        average_order_value
    ]
})

kpi_summary
```

```python
kpi_summary['Value'] = kpi_summary['Value'].round(2)

kpi_summary
```

```python
kpi_display = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Orders',
        'Total Customers',
        'Total Products',
        'Average Order Value'
    ],
    'Value': [
        f"${total_sales:,.2f}",
        f"{total_orders:,}",
        f"{total_customers:,}",
        f"{total_products:,}",
        f"${average_order_value:,.2f}"
    ]
})

kpi_display
```

## Regional Sales Analysis

Regional analysis helps identify how sales are distributed across different geographical regions. This can help the business understand which regions are performing strongly and where there may be opportunities for improvement.

```python
regional_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

regional_sales
```

```python
regional_sales_table = regional_sales.reset_index()

regional_sales_table.columns = ['Region', 'Total Sales']

regional_sales_table['Total Sales'] = regional_sales_table['Total Sales'].round(2)

regional_sales_table
```

```python
highest_sales_region = regional_sales.idxmax()
highest_sales_value = regional_sales.max()

highest_sales_region, highest_sales_value
```

```python
lowest_sales_region = regional_sales.idxmin()
lowest_sales_value = regional_sales.min()

lowest_sales_region, lowest_sales_value
```

```python
regional_sales_percentage = (regional_sales / total_sales * 100).round(2)

regional_sales_percentage
```

```python
regional_summary = pd.DataFrame({
    'Region': regional_sales.index,
    'Total Sales': regional_sales.values.round(2),
    'Sales Contribution (%)': regional_sales_percentage.values
})

regional_summary
```

### Sales by Region

The following chart compares total sales across the four regions and helps highlight differences in regional sales performance.

```python
plt.figure(figsize=(8, 5))

plt.bar(regional_summary['Region'], regional_summary['Total Sales'])

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(
    regional_summary['Region'],
    regional_summary['Total Sales']
)

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f'${value:,.0f}',
        ha='center',
        va='bottom'
    )

plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

## Sales Trend Analysis

Time-based analysis helps identify changes in sales performance over different periods. Studying yearly and monthly sales trends can help the business understand growth patterns, seasonal changes, and periods of stronger or weaker performance.

```python
df['Order_Date'].dtype
```

```python
df['Order_Year'] = df['Order_Date'].dt.year
```

```python
yearly_sales = df.groupby('Order_Year')['Sales'].sum().sort_index()

yearly_sales
```

```python
yearly_summary = yearly_sales.reset_index()

yearly_summary.columns = ['Year', 'Total Sales']

yearly_summary['Total Sales'] = yearly_summary['Total Sales'].round(2)

yearly_summary
```

```python
yearly_summary['YoY Growth (%)'] = (
    yearly_summary['Total Sales'].pct_change() * 100
).round(2)

yearly_summary
```

### Yearly Sales Trend

The following chart shows the change in total sales from 2015 to 2018. It helps visualize the overall direction of sales performance over the four-year period.

```python
plt.figure(figsize=(9, 5))

plt.plot(
    yearly_summary['Year'],
    yearly_summary['Total Sales'],
    marker='o'
)

plt.title('Yearly Sales Trend')
plt.xlabel('Year')
plt.ylabel('Total Sales')

for x, y in zip(yearly_summary['Year'], yearly_summary['Total Sales']):
    plt.text(
        x,
        y,
        f'${y:,.0f}',
        ha='center',
        va='bottom'
    )

plt.xticks(yearly_summary['Year'])
plt.tight_layout()

plt.show()
```

### Yearly Sales Insights

- Sales decreased by 4.26% in 2016 compared with 2015.
- Sales increased significantly by 30.64% in 2017 compared with 2016.
- Sales continued to grow in 2018, increasing by 20.30% compared with 2017.
- 2018 recorded the highest annual sales of approximately $722,052.02.
- Overall, the data shows a strong upward sales trend after the decline observed in 2016.

### Business Recommendation

The strong growth observed in 2017 and 2018 suggests that the business should examine the factors associated with this improvement and continue strategies that supported the growth. The decline in 2016 should also be investigated to understand what may have contributed to the temporary reduction in sales.

## Sales and Order Analysis

Sales and order analysis examines the number of unique orders and the average sales generated per order. This provides a broader understanding of the overall sales pattern of the business.

```python
total_orders = df['Order_ID'].nunique()

print(f"Total Unique Orders: {total_orders}")
```
